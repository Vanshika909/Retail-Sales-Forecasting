```python
df_clean['Year'] = df_clean['Order Date'].dt.year

df_clean['Month'] = df_clean['Order Date'].dt.month_name()

df_clean['Month_Number'] = df_clean['Order Date'].dt.month

df_clean['Quarter'] = df_clean['Order Date'].dt.quarter
```

```python
import matplotlib.pyplot as plt

category_sales = df_clean.groupby('Category')['Sales'].sum()

category_sales.plot(kind='bar', figsize=(8,5), color='Blue')

plt.title('Category-wise Sales')
plt.xlabel('Category')
plt.ylabel('Total Sales')
plt.xticks(rotation=0)
plt.show()
```

```python
region_sales = df_clean.groupby('Region')['Sales'].sum()

region_sales.plot(kind='bar', figsize=(8,5))

plt.title('Region-wise Sales')
plt.xlabel('Region')
plt.ylabel('Total Sales')
plt.xticks(rotation=0)
plt.show()
```

```python
segment_sales = df_clean.groupby('Segment')['Sales'].sum()

segment_sales.plot(kind='pie', autopct='%1.1f%%', figsize=(6,6))

plt.title('Segment-wise Sales Contribution')
plt.ylabel('')
plt.show()
```

```python
monthly_sales = df_clean.groupby('Month_Number')['Sales'].sum()

monthly_sales.plot(kind='line', marker='o', figsize=(10,5))

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.grid(True)
plt.show()
```

```python
profit_category = df_clean.groupby('Category')['Profit'].sum()

profit_category.plot(kind='bar', figsize=(8,5))

plt.title('Profit by Category')
plt.xlabel('Category')
plt.ylabel('Profit')
plt.xticks(rotation=0)
plt.show()
```

```python
subcat_sales = df_clean.groupby('Sub-Category')['Sales'].sum().sort_values(ascending=False)

subcat_sales.plot(kind='bar', figsize=(12,6))
plt.title('Sub-Category Wise Sales')
plt.xlabel('Sub-Category')
plt.ylabel('Sales')
plt.xticks(rotation=45)
plt.show()
```

```python
top_products = df_clean.groupby('Product Name')['Sales'].sum().sort_values(ascending=False).head(10)

top_products.plot(kind='bar', figsize=(12,6))
plt.title('Top 10 Products by Sales')
plt.ylabel('Sales')
plt.xticks(rotation=90)
plt.show()
```

```python
top_customers = df_clean.groupby('Customer Name')['Sales'].sum().sort_values(ascending=False).head(10)

top_customers.plot(kind='bar', figsize=(12,6))
plt.title('Top 10 Customers by Sales')
plt.ylabel('Sales')
plt.xticks(rotation=90)
plt.show()
```

```python
state_sales = df_clean.groupby('State')['Sales'].sum().sort_values(ascending=False).head(10)

state_sales.plot(kind='bar', figsize=(12,6))
plt.title('Top 10 States by Sales')
plt.ylabel('Sales')
plt.xticks(rotation=45)
plt.show()
```

```python
import seaborn as sns

corr = df_clean[['Sales','Profit','Quantity','Discount']].corr()

sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title('Correlation Matrix')
plt.show()
```

```python
sales_ts = df_clean.groupby('Order Date')['Sales'].sum().reset_index()
sales_ts.columns = ['ds', 'y']
sales_ts.head()
```

```python
plt.figure(figsize=(12,5))
plt.plot(sales_ts['ds'], sales_ts['y'])
plt.title('Daily Sales Trend')
plt.xlabel('Date')
plt.ylabel('Sales')
plt.grid(True)
plt.show()
```

```python
from prophet import Prophet
```

```python
df.head()
```

```python
sales_ts.head()
```

```python
model = Prophet()
model.fit(sales_ts)
```

```python
future = model.make_future_dataframe(periods=90)

future.tail()
```

```python
forecast = model.predict(future)
forecast[['ds', 'yhat', 'yhat_lower', 'yhat_upper']].tail()
```

```python
fig = model.plot(forecast)

plt.title('Retail Sales Forecast - Next 90 Days')
plt.xlabel('Date')
plt.ylabel('Predicted Sales')
plt.show()
```

```python
future_predictions = forecast[['ds', 'yhat']].tail(90)

future_predictions.head()
```

```python
future_predictions.to_csv(
    '../data/future_sales_predictions.csv',
    index=False
)
```
